# **Qu'est ce qu'un CNN ?**

Les **Réseaux de Neurones Convolutifs (CNN)** se sont révélés particulièrement performants pour les tâches de classification d’images grâce à leur capacité à apprendre automatiquement des caractéristiques hiérarchiques à partir des pixels bruts.

Grâce à l’utilisation de :

* **Couches de Convolution (Convolutional Layers)**, qui détectent les motifs visuels tels que les contours, les textures et les formes ;
* **Couches de Pooling (Pooling Layers)**, qui réduisent la dimension des données tout en conservant les informations importantes ;
* **Couches Entièrement Connectées (Fully Connected Layers)**, qui réalisent la classification finale ;

les CNN sont capables de capturer des motifs complexes et les relations spatiales présentes dans les images. Cette capacité les rend particulièrement adaptés aux problèmes de reconnaissance visuelle tels que la classification des images du jeu de données **CIFAR-10**.

### **Préparation des données**

Avant d'utiliser les données pour entraîner notre modèle, nous appliquons une série de transformations afin d'augmenter et de prétraiter les images avant de les fournir à notre réseau de neurones convolutif (CNN). Ces transformations sont implémentées à l'aide du module `transforms` de PyTorch et sont regroupées dans un pipeline grâce à `transforms.Compose`.

1. **Random Horizontal Flip (`transforms.RandomHorizontalFlip(p=0.5)`)**

   Cette transformation retourne aléatoirement l'image horizontalement avec une probabilité de 0,5. Elle permet d'introduire davantage de diversité dans les données d'entraînement en simulant différentes orientations des objets présents dans les images.

2. **Random Rotation (`transforms.RandomRotation(20)`)**

   Cette transformation effectue une rotation aléatoire de l'image jusqu'à un maximum de 20 degrés. Elle aide le modèle à devenir moins sensible à l'orientation des objets dans les images.

3. **Color Jitter (`transforms.ColorJitter(brightness=0.1, contrast=0.1, saturation=0.1)`)**

   Cette transformation modifie aléatoirement la luminosité, le contraste et la saturation de l'image. Elle rend le modèle plus robuste face aux variations d'éclairage et de couleur.

4. **Random Adjust Sharpness (`transforms.RandomAdjustSharpness(sharpness_factor=2, p=0.2)`)**

   Cette transformation ajuste aléatoirement la netteté de l'image avec une probabilité de 0,2. Elle introduit des variations dans la qualité visuelle des images, ce qui améliore la capacité du modèle à généraliser.

5. **To Tensor (`transforms.ToTensor()`)**

   Cette transformation convertit l'image en un tenseur PyTorch et met les valeurs des pixels à l'échelle dans l'intervalle [0.0, 1.0]. Elle prépare ainsi les données pour les calculs basés sur les tenseurs dans PyTorch.

6. **Normalization (`transforms.Normalize(mean, std)`)**

   Cette transformation normalise l'image à l'aide des valeurs de moyenne (*mean*) et d'écart-type (*std*) spécifiées. Ces valeurs sont largement utilisées dans la recherche car elles favorisent une convergence plus rapide de l'entraînement. La normalisation contribue également à stabiliser le processus d'apprentissage en plaçant les caractéristiques d'entrée sur une échelle comparable.

7. **Random Erasing (`transforms.RandomErasing(p=0.75, scale=(0.02, 0.1), value=1.0, inplace=False)`)**

   Cette transformation efface aléatoirement des régions rectangulaires de l'image avec une probabilité de 0,75. Elle agit comme une technique de régularisation et aide à réduire le surapprentissage (*overfitting*) en obligeant le modèle à apprendre des caractéristiques plus robustes.

```python
## Importing modules ##

# Import PyTorch
import torch
from torch import nn
import torch.nn.functional as F
from torch.utils.data import DataLoader

# Import torchvision
import torchvision
from torchvision import datasets
from torchvision import transforms

# for file system
import os

# Import matplotlib for visualization
import matplotlib.pyplot as plt

# for model evaluation
from sklearn.metrics import confusion_matrix
import pandas as pd
import seaborn as sn
import numpy as np

# for randimization
import random

# type
from typing import List , Dict , Tuple
```

```python
# Setup device agnostic code
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

```python
# transforms object to use in data augmentation

mean, std = [0.4914, 0.4822, 0.4465], [0.247, 0.243, 0.261]

data_transform = transforms.Compose([
    # Flip the images randomly on the horizontal
    transforms.RandomHorizontalFlip(p=0.5),
    # Randomly rotate some images by 20 degrees
    transforms.RandomRotation(20),
    # Randomly adjust color jitter of the images
    transforms.ColorJitter(brightness = 0.1,contrast = 0.1,saturation = 0.1),
    # Randomly adjust sharpness
    transforms.RandomAdjustSharpness(sharpness_factor = 2,p = 0.2),
    # Turn the image into a torch.Tensor
    transforms.ToTensor() ,
    #randomly erase a pixel
    transforms.Normalize(mean, std),
    transforms.RandomErasing(p=0.75,scale=(0.02, 0.1),value=1.0, inplace=False)
])
```

```python
# Setup training data

train_data = datasets.CIFAR10(
    root="data", # where to download data to?
    train=True, # get training data
    download=True, # download data if it doesn't exist on disk
    transform=data_transform, # images come as PIL format, we want to turn into Torch tensors
    target_transform=None # you can transform labels as well
)

# Setup testing data

test_data = datasets.CIFAR10(
    root="data",
    train=False, # get test data
    download=True,
    transform=data_transform
)
```

```python
#see classes

class_names = train_data.classes

class_names
```

```python
# look into the data we've downloaded

rand_idx = random.sample(range(len(train_data)),k=16)

plt.figure(figsize=(10,10))

for i , idx in enumerate(rand_idx):

  img , label = train_data[idx]

  #the image tensor's range is not between 0 and 1,so we have to temporarily scale the tensor values into range 0 and 1 to prevent error.
  img = (img - img.min()) / (img.max() - img.min())

  img_class = class_names[label]

  plt.subplot(4,4,i+1)
  plt.imshow(img.permute(1,2,0))
  plt.title(f"Class : {img_class}",fontsize=10)
  plt.axis(False)
```

```python
#load data into dataloader

NUM_WORKERS = os.cpu_count()
BATCH_SIZE = 800

train_dataloader = DataLoader(dataset=train_data,
                              batch_size=BATCH_SIZE,
                              num_workers=NUM_WORKERS,
                              shuffle=True)

test_dataloader = DataLoader(dataset=test_data,
                             batch_size=BATCH_SIZE,
                             num_workers=NUM_WORKERS,
                             shuffle=False)

```

```python
# train loop function
def train_step(model:nn.Module,
               dataloader:torch.utils.data.DataLoader,
               loss_fn:nn.Module,
               optimizer:torch.optim.Optimizer,
               scheduler:torch.optim.lr_scheduler = None,
               grad_clip:float=None):

  model.train()

  train_loss , train_acc = 0 , 0

  for batch , (X,y) in enumerate(dataloader):

    X , y = X.to(device) , y.to(device)
    y_pred = model(X)
    loss = loss_fn(y_pred,y)
    train_loss += loss.item()
    optimizer.zero_grad()
    loss.backward()

    if grad_clip:
      nn.utils.clip_grad_value_(model.parameters(), grad_clip)

    optimizer.step()
    y_pred_class = torch.argmax(y_pred,dim=1)
    train_acc += (y_pred_class == y).sum().item() / len(y)

  train_loss /= len(dataloader)
  train_acc /= len(dataloader)

  if scheduler is not None:
    scheduler.step(train_loss)


  return train_loss , train_acc
```

```python
# test loop function
def test_step(model:nn.Module,
              dataloader:torch.utils.data.DataLoader,
              loss_fn:nn.Module):

  model.eval()

  test_loss , test_acc = 0 , 0

  with torch.inference_mode():
    for batch , (X,y) in enumerate(dataloader):
      X , y = X.to(device) , y.to(device)
      test_pred_logits = model(X)
      loss = loss_fn(test_pred_logits,y)
      test_loss += loss.item()
      test_pred_labels = torch.argmax(test_pred_logits,dim=1)
      test_acc += (test_pred_labels == y).sum().item() / len(y)

    test_loss /= len(dataloader)
    test_acc /= len(dataloader)

  return test_loss , test_acc
```

```python
#Creating a train() function to combine train_step() and test_step()

from tqdm.auto import tqdm

def train(model: torch.nn.Module,
          train_dataloader: torch.utils.data.DataLoader,
          test_dataloader: torch.utils.data.DataLoader,
          optimizer: torch.optim.Optimizer,
          scheduler:torch.optim.lr_scheduler,
          grad_clip:float=None,
          loss_fn: torch.nn.Module = nn.CrossEntropyLoss(),
          epochs: int = 10):

    results = {"train_loss": [],
        "train_acc": [],
        "test_loss": [],
        "test_acc": []
    }

    for epoch in tqdm(range(epochs)):
        train_loss, train_acc = train_step(model=model,
                                           dataloader=train_dataloader,
                                           loss_fn=loss_fn,
                                           optimizer=optimizer,
                                           scheduler=scheduler,
                                           grad_clip=grad_clip)
        test_loss, test_acc = test_step(model=model,
            dataloader=test_dataloader,
            loss_fn=loss_fn)


        print(
            f"Epoch: {epoch+1} | "
            f"train_loss: {train_loss:.4f} | "
            f"train_acc: {train_acc:.4f} | "
            f"test_loss: {test_loss:.4f} | "
            f"test_acc: {test_acc:.4f}"
        )


        results["train_loss"].append(train_loss)
        results["train_acc"].append(train_acc)
        results["test_loss"].append(test_loss)
        results["test_acc"].append(test_acc)

    return results
```

```python
# function to plot loss & accuracy curve

def plot_loss_curves(results: Dict[str, List[float]]):


    # Get the loss values of the results dictionary (training and test)
    loss = results['train_loss']
    test_loss = results['test_loss']

    # Get the accuracy values of the results dictionary (training and test)
    accuracy = results['train_acc']
    test_accuracy = results['test_acc']

    # Figure out how many epochs there were
    epochs = range(len(results['train_loss']))

    # Setup a plot
    plt.figure(figsize=(15, 7))

    # Plot loss
    plt.subplot(1, 2, 1)
    plt.plot(epochs, loss, label='train_loss')
    plt.plot(epochs, test_loss, label='test_loss')
    plt.title('Loss')
    plt.xlabel('Epochs')
    plt.grid()
    plt.legend()


    # Plot accuracy
    plt.subplot(1, 2, 2)
    plt.plot(epochs, accuracy, label='train_accuracy')
    plt.plot(epochs, test_accuracy, label='test_accuracy')
    plt.title('Accuracy')
    plt.xlabel('Epochs')
    plt.grid()
    plt.legend()
```

## **Modèle CNN**

Nous allons maintenant définir l'architecture de notre Réseau de Neurones Convolutif (CNN). Ce modèle est conçu pour la classification d'images sur le jeu de données CIFAR-10. Il comprend les couches suivantes :

1.  **Bloc Convolutif 1**:
    *   `Conv2d`: Applique 32 filtres de taille 3x3 avec un pas de 1 et un padding de 1. L'entrée est une image couleur (3 canaux).
    *   `ReLU`: Fonction d'activation non linéaire.
    *   `Conv2d`: Applique 32 filtres de taille 3x3 avec un pas de 1 et un padding de 1.
    *   `ReLU`: Fonction d'activation.
    *   `MaxPool2d`: Réduit la dimension spatiale de 2x2 avec un pas de 2.
    *   `Dropout`: Aide à prévenir le surapprentissage en désactivant aléatoirement 25% des neurones.

2.  **Bloc Convolutif 2**:
    *   `Conv2d`: Applique 64 filtres de taille 3x3 avec un pas de 1 et un padding de 1.
    *   `ReLU`: Fonction d'activation.
    *   `Conv2d`: Applique 64 filtres de taille 3x3 avec un pas de 1 et un padding de 1.
    *   `ReLU`: Fonction d'activation.
    *   `MaxPool2d`: Réduit la dimension spatiale de 2x2 avec un pas de 2.
    *   `Dropout`: Désactive aléatoirement 25% des neurones.

3.  **Bloc Convolutif 3**:
    *   `Conv2d`: Applique 128 filtres de taille 3x3 avec un pas de 1 et un padding de 1.
    *   `ReLU`: Fonction d'activation.
    *   `Conv2d`: Applique 128 filtres de taille 3x3 avec un pas de 1 et un padding de 1.
    *   `ReLU`: Fonction d'activation.
    *   `MaxPool2d`: Réduit la dimension spatiale de 2x2 avec un pas de 2.
    *   `Dropout`: Désactive aléatoirement 25% des neurones.

4.  **Couches Linéaires (Fully Connected Layers)**:
    *   `Flatten`: Aplatit la sortie des couches convolutives en un vecteur.
    *   `Linear`: Première couche entièrement connectée avec une réduction du nombre de caractéristiques.
    *   `ReLU`: Fonction d'activation.
    *   `Dropout`: Désactive aléatoirement 50% des neurones.
    *   `Linear`: Deuxième couche entièrement connectée qui mappe les caractéristiques aux 10 classes de sortie (CIFAR-10).

```python
import torch
from torch import nn
# Create CNN model with `nn.Module`
class CNN_Model(nn.Module):
    def __init__(self, input_shape: int, hidden_units: int, output_shape: int):
        super().__init__()
        self.block_1 = nn.Sequential(
            nn.Conv2d(in_channels=input_shape,
                      out_channels=hidden_units,
                      kernel_size=3,
                      stride=1,
                      padding=1),
            nn.ReLU(),
            nn.Conv2d(in_channels=hidden_units,
                      out_channels=hidden_units,
                      kernel_size=3,
                      stride=1,
                      padding=1),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2,
                         stride=2),
            nn.Dropout(p=0.25)
        )
        self.block_2 = nn.Sequential(
            nn.Conv2d(hidden_units, hidden_units*2, 3, padding=1),
            nn.ReLU(),
            nn.Conv2d(hidden_units*2, hidden_units*2, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Dropout(p=0.25)
        )
        self.block_3 = nn.Sequential(
            nn.Conv2d(hidden_units*2, hidden_units*4, 3, padding=1),
            nn.ReLU(),
            nn.Conv2d(hidden_units*4, hidden_units*4, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Dropout(p=0.25)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(in_features=hidden_units*4*4*4,
                      out_features=output_shape*2),
            nn.ReLU(),
            nn.Dropout(p=0.5),
            nn.Linear(in_features=output_shape*2,
                      out_features=output_shape)
        )

    def forward(self, x: torch.Tensor):
        x = self.block_1(x)
        x = self.block_2(x)
        x = self.block_3(x)
        x = self.classifier(x)
        return x
```

### **Instanciation du modèle et entraînement**

Nous allons maintenant instancier notre modèle CNN, définir la fonction de perte (`CrossEntropyLoss`), l'optimiseur (`Adam`) et le scheduler (`ReduceLROnPlateau`). Ensuite, nous lancerons l'entraînement du modèle pour un nombre défini d'époques.

```python
# Create an instance of the model
model_0 = CNN_Model(input_shape=3, # number of color channels (3 for RGB)
                     hidden_units=32,
                     output_shape=len(class_names)).to(device)

# Setup loss function and optimizer
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model_0.parameters(), lr=0.001)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.1, patience=5)

# Train the model
epochs = 20
model_0_results = train(model=model_0,
                        train_dataloader=train_dataloader,
                        test_dataloader=test_dataloader,
                        optimizer=optimizer,
                        scheduler=scheduler,
                        loss_fn=loss_fn,
                        epochs=epochs)

# Plot the loss curves
plot_loss_curves(model_0_results)
```

# *Modele AlexNet*

AlexNet est une architecture de réseau de neurones convolutifs profond proposée par Alex Krizhevsky, Ilya Sutskever et Geoffrey Hinton en 2012. Elle a révolutionné le domaine de la vision par ordinateur en remportant la compétition ImageNet et en démontrant l'efficacité des réseaux de neurones profonds pour la classification d'images, marquant ainsi le début de l'ère du deep learning moderne.

## *Contexte*

Avant l'émergence d'AlexNet, les méthodes traditionnelles de vision par ordinateur basées sur les caractéristiques manuelles (hand-crafted features) dominaient le domaine. Cependant, ces approches avaient des limitations significatives dans leur capacité à généraliser et à traiter des images complexes et variées. Le défi principal résidait dans le manque de puissance de calcul suffisante pour entraîner des réseaux profonds sur de grands ensembles de données.

AlexNet a surmonté ces obstacles en tirant parti de plusieurs innovations techniques clés : l'utilisation de GPU pour accélérer considérablement l'entraînement, l'application de la régularisation par dropout pour réduire le surapprentissage, et l'utilisation de la fonction d'activation ReLU qui s'est avérée plus efficace que les fonctions d'activation traditionnelles.

## *Architecture*

AlexNet est composée de cinq couches de convolution suivies de trois couches entièrement connectées. La première couche de convolution traite les images d'entrée (224 x 224 pixels) avec 96 filtres de taille 11 x 11, stride de 4. Les couches suivantes réduisent progressivement les dimensions spatiales tout en augmentant la profondeur des caractéristiques extraites.

Entre les couches de convolution, AlexNet emploie des opérations de pooling en max pour réduire les dimensions spatiales. Les couches finales sont des couches entièrement connectées comportant 4096, 4096 et 1000 neurones respectivement, permettant la classification dans les 1000 catégories d'ImageNet.

Les innovations architecturales incluent l'utilisation généralisée de ReLU au lieu de fonctions de saturation, l'implémentation du dropout dans les couches entièrement connectées pour réduire le surapprentissage, et la normalisation locale des réponses (Local Response Normalization) pour améliorer la généralisation. Ces éléments ont contribué à faire d'AlexNet un modèle fondamental qui a inspiré toutes les architectures plus profondes qui ont suivi.

```python
import torch
import torch.nn as nn

class AlexNet(nn.Module):

    def conv_block(self,
                   in_channels,
                   out_channels,
                   kernel_size,
                   stride=1,
                   padding=0,
                   pool=False):

        layers = [
            nn.Conv2d(
                in_channels,
                out_channels,
                kernel_size=kernel_size,
                stride=stride,
                padding=padding
            ),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True)
        ]

        if pool:
            layers.append(nn.MaxPool2d(kernel_size=2, stride=2))

        return nn.Sequential(*layers)

    def __init__(self, in_channels=3, num_classes=10):
        super().__init__()

        self.features = nn.Sequential(

            # 32x32 -> 16x16
            self.conv_block(
                in_channels,
                96,
                kernel_size=3,
                stride=1,
                padding=1,
                pool=True
            ),

            # 16x16 -> 8x8
            self.conv_block(
                96,
                256,
                kernel_size=3,
                stride=1,
                padding=1,
                pool=True
            ),

            # 8x8
            self.conv_block(
                256,
                384,
                kernel_size=3,
                stride=1,
                padding=1
            ),

            # 8x8
            self.conv_block(
                384,
                384,
                kernel_size=3,
                stride=1,
                padding=1
            ),

            # 8x8 -> 4x4
            self.conv_block(
                384,
                256,
                kernel_size=3,
                stride=1,
                padding=1,
                pool=True
            )
        )

        self.classifier = nn.Sequential(
            nn.AdaptiveAvgPool2d((1, 1)),
            nn.Flatten(),

            nn.Linear(256, 4096),
            nn.ReLU(inplace=True),
            nn.Dropout(0.5),

            nn.Linear(4096, 4096),
            nn.ReLU(inplace=True),
            nn.Dropout(0.5),

            nn.Linear(4096, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x
```

# **Modèle VGG**

Le modèle **VGG** (*Visual Geometry Group*) est une architecture de réseau de neurones convolutif profond (*Deep Convolutional Neural Network*) proposée par des chercheurs de l'University of Oxford. Il est devenu populaire grâce à sa simplicité et à son efficacité dans les tâches de reconnaissance d'images, notamment lors du concours **ILSVRC (ImageNet Large Scale Visual Recognition Challenge)**.

L'architecture VGG est composée de plusieurs **couches de convolution** suivies de **couches de max-pooling**, puis de **couches entièrement connectées** à la fin du réseau pour effectuer la classification. Les principales caractéristiques de l'architecture VGG sont les suivantes :

* **Couches de convolution (Convolutional Layers) :** Les couches de convolution utilisent de petits filtres de taille **3 × 3**, avec un pas (*stride*) de **1** et un remplissage (*zero-padding*) permettant de conserver la résolution spatiale des cartes de caractéristiques (*feature maps*). La profondeur des cartes de caractéristiques augmente progressivement à mesure que l'on avance dans le réseau.

* **Couches de max-pooling (Max Pooling Layers) :** Après chaque groupe de couches de convolution, des couches de max-pooling utilisant une fenêtre de **2 × 2** et un pas de **2** sont appliquées afin de réduire les dimensions spatiales des cartes de caractéristiques tout en conservant les informations les plus importantes.

* **Couches entièrement connectées (Fully Connected Layers) :** Les dernières couches de l'architecture VGG sont des couches entièrement connectées, également appelées **couches denses (Dense Layers)**. Elles prennent en entrée les cartes de caractéristiques aplaties (*flattened feature maps*) issues des couches de convolution et produisent les scores finaux (*logits*) utilisés pour la classification.

La classe **VGG** présentée dans ce projet définit une architecture simplifiée de réseau de neurones convolutif (CNN), inspirée du modèle VGG original, mais comportant un nombre réduit de couches et de paramètres afin de faciliter son entraînement et son utilisation sur le jeu de données CIFAR-10.

image.png

```python
#build a model

class TinyVGG(nn.Module):
  def __init__(self, input_shape: int, output_shape: int) -> None:
    super().__init__()
    self.conv_block_1 = nn.Sequential(
        nn.Conv2d(in_channels=input_shape,
                  out_channels=32,
                  kernel_size=3,
                  stride=1,
                  padding=1),
        nn.ReLU(),
        nn.Conv2d(in_channels=32,
                  out_channels=32,
                  kernel_size=3,
                  stride=1,
                  padding=1),
        nn.ReLU(),
        nn.MaxPool2d(kernel_size=2,
                     stride=2)
    )
    self.conv_block_2 = nn.Sequential(
        nn.Conv2d(in_channels=32,
                  out_channels=64,
                  kernel_size=3,
                  stride=1,
                  padding=1),
        nn.ReLU(),
        nn.Conv2d(in_channels=64,
                  out_channels=64,
                  kernel_size=3,
                  stride=1,
                  padding=1),
        nn.ReLU(),
        nn.MaxPool2d(kernel_size=2,
                     stride=2)
    )

    self.classifier = nn.Sequential(
        nn.Flatten(),
        nn.Linear(in_features=64*8*8,out_features=2048),
        nn.ReLU(),
        nn.Dropout(p=0.5), #add dropout layer to decrease overfitting
        nn.Linear(in_features=2048,out_features=1024),
        nn.ReLU(),
        nn.Dropout(p=0.5),
        nn.Linear(in_features=1024,out_features=output_shape)
    )



  def forward(self, x: torch.Tensor):
        x = self.conv_block_1(x)
        x = self.conv_block_2(x)
        x = self.classifier(x)
        return x
```

# **Entraînement du modèle**

## Explication des hyperparamètres

### 1. Nombre d'époques (Epoch)

* `NUM_EPOCHS` représente le nombre de passages complets du jeu de données d'entraînement à travers le réseau. Il doit être suffisamment élevé pour permettre l'apprentissage sans provoquer de surapprentissage.

### 2. Taux d'apprentissage (Learning Rate)

* `learning_rate` contrôle la vitesse d'apprentissage du modèle. Une valeur élevée accélère l'entraînement mais peut réduire la stabilité, tandis qu'une valeur faible favorise une convergence plus précise.

### 3. Décroissance des poids (Weight Decay)

* `weight_decay` applique une régularisation L2 pour limiter le surapprentissage en pénalisant les poids trop importants.

### 4. Limitation des gradients (Gradient Clipping)

* `grad_clip` limite la valeur des gradients afin d'éviter leur explosion et de stabiliser l'entraînement.

### 5. Fonction de perte et optimiseur

* `loss_fn = nn.CrossEntropyLoss()` : fonction de perte adaptée aux problèmes de classification multiclasse.
* `optimizer = torch.optim.Adam(...)` : optimiseur efficace qui ajuste automatiquement les paramètres du modèle durant l'apprentissage.

### 6. Planificateur du taux d'apprentissage (Learning Rate Scheduler)

* `scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(...)` : réduit automatiquement le taux d'apprentissage lorsque les performances du modèle cessent de s'améliorer, afin de favoriser une meilleure convergence.

# *Modele ResNet*

ResNet, abrégé pour Residual Network, est une architecture de réseau de neurones convolutifs profond proposée par des chercheurs de Microsoft Research. Elle a introduit un concept révolutionnaire appelé apprentissage résiduel, qui résout le problème de la disparition des gradients dans les réseaux profonds et permet l'entraînement de réseaux de neurones très profonds avec des centaines ou même des milliers de couches.

## *Contexte*

Les réseaux de neurones profonds traditionnels font face à des défis dans l'entraînement d'architectures très profondes en raison du problème de disparition des gradients. À mesure que la profondeur du réseau augmente, les gradients peuvent devenir extrêmement petits lors de la rétropropagation, ce qui rend difficile l'apprentissage efficace du réseau.

ResNet aborde ce problème en introduisant des connexions résiduelles, aussi appelées connexions de saut (skip connections), qui permettent au réseau d'apprendre des mappages résiduels au lieu d'apprendre directement le mappage sous-jacent souhaité. Cela contribue à atténuer le problème de dégradation causé par les réseaux profonds et permet l'entraînement d'architectures plus profondes.

## *Architecture*

Le bloc de construction fondamental de l'architecture ResNet est le bloc résiduel, qui se compose de deux couches de convolution avec normalisation par batch et fonctions d'activation ReLU, ainsi qu'une connexion de raccourci (shortcut) qui contourne une ou plusieurs couches.

La connexion de raccourci ajoute l'entrée originale à la sortie du bloc résiduel, permettant au réseau d'apprendre des mappages résiduels. Cela permet aux gradients de circuler plus facilement à travers le réseau, facilitant ainsi l'entraînement d'architectures plus profondes.

Les architectures ResNet empilent généralement plusieurs blocs résiduels pour former des réseaux plus profonds. Différentes variantes de ResNet, telles que ResNet-18, ResNet-34, ResNet-50, ResNet-101 et ResNet-152, varient en termes de nombre de couches et du nombre de blocs résiduels par couche.


image.png

```python
# create Resnet model class

class ResNet9(nn.Module):
    def conv_block(self,in_channels, out_channels, pool=False):
      layers = [nn.Conv2d(in_channels, out_channels, kernel_size=3, padding=1),
                nn.BatchNorm2d(out_channels),
                nn.ReLU(inplace=True)]
      if pool: layers.append(nn.MaxPool2d(2))
      return nn.Sequential(*layers)

    def __init__(self, in_channels, num_classes):
        super().__init__()

        self.conv1 = self.conv_block(in_channels, 64)
        self.conv2 = self.conv_block(64, 128, pool=True)
        self.res1 = nn.Sequential(self.conv_block(128, 128), self.conv_block(128, 128))

        self.conv3 = self.conv_block(128, 256, pool=True)
        self.conv4 = self.conv_block(256, 512, pool=True)
        self.res2 = nn.Sequential(self.conv_block(512, 512), self.conv_block(512, 512))

        self.classifier = nn.Sequential(nn.MaxPool2d(4),
                                        nn.Flatten(),
                                        nn.Linear(512, num_classes))

    def forward(self, x):
        out = self.conv1(x)
        out = self.conv2(out)
        out = self.res1(out) + out
        out = self.conv3(out)
        out = self.conv4(out)
        out = self.res2(out) + out
        out = self.classifier(out)
        return out
```

# *Modele EfficientNet*

EfficientNet est une famille d'architectures de réseaux de neurones convolutifs proposée par Mingxue Tan et Quoc V. Le en 2019. Elle introduit une approche novatrice appelée compound scaling qui équilibre de manière optimale la profondeur, la largeur et la résolution des images d'entrée. EfficientNet a établi un nouvel équilibre entre la précision et l'efficacité computationnelle, permettant des modèles plus compacts et plus rapides sans compromettre les performances.

## *Contexte*

Bien que les réseaux profonds comme ResNet et VGGNet aient obtenu des résultats impressionnants sur ImageNet, ces architectures consomment énormément de ressources computationnelles et possèdent des millions de paramètres, les rendant impraticables pour les applications mobiles et les appareils avec des ressources limitées. Le défi principal a été de trouver un compromis optimal entre la performance du modèle, la latence d'inférence et l'utilisation de la mémoire.

EfficientNet résout ce problème en proposant une méthode systématique pour scalabiliser les réseaux de neurones convolutifs. Plutôt que de simplement augmenter la profondeur (comme dans ResNet) ou la largeur (comme dans les architectures larges), EfficientNet utilise le compound scaling qui ajuste simultanément et proportionnellement trois dimensions : la profondeur du réseau, la largeur des couches, et la résolution des images d'entrée selon un ensemble de coefficients optimisés.

## *Architecture*

EfficientNet est basée sur une architecture de base appelée EfficientNet-B0, construite à partir de blocs appelés Mobile Inverted Bottleneck (MBConv). Ces blocs combinent les convolutions dépthwise et pointwise pour réduire le nombre de paramètres tout en maintenant la capacité d'apprentissage.

La famille EfficientNet comprend huit variantes, de B0 à B7, où chaque variante est obtenue en appliquant le compound scaling à différents degrés. Pour chaque nouvelle variante, la profondeur est augmentée par un facteur α, la largeur par un facteur β, et la résolution d'entrée par un facteur γ, où ces coefficients sont déterminés de manière à maximiser l'efficacité énergétique.

Les caractéristiques clés incluent l'utilisation de squeeze-and-excitation modules pour affiner les poids des canaux, l'emploi d'une normalisation batch efficace, et une stratégie d'entraînement appelée progressive image size augmentation qui augmente graduellement la résolution d'entrée. Cette approche permet aux modèles EfficientNet d'atteindre une meilleure précision avec significativement moins de paramètres et d'opérations que les architectures concurrentes, les rendant particulièrement adaptés pour les applications où les ressources computationnelles sont limitées.

```python
import torch
import torch.nn as nn


class SEBlock(nn.Module):
    def __init__(self, channels, reduction=4):
        super().__init__()

        self.se = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Conv2d(channels, channels // reduction, kernel_size=1),
            nn.SiLU(),
            nn.Conv2d(channels // reduction, channels, kernel_size=1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return x * self.se(x)


class MBConv(nn.Module):
    def __init__(self,
                 in_channels,
                 out_channels,
                 expand_ratio=6,
                 stride=1):

        super().__init__()

        hidden_dim = in_channels * expand_ratio
        self.use_residual = (
            stride == 1 and in_channels == out_channels
        )

        layers = []

        # Expansion
        if expand_ratio != 1:
            layers.extend([
                nn.Conv2d(in_channels, hidden_dim, 1, bias=False),
                nn.BatchNorm2d(hidden_dim),
                nn.SiLU()
            ])
        else:
            hidden_dim = in_channels

        # Depthwise
        layers.extend([
            nn.Conv2d(
                hidden_dim,
                hidden_dim,
                kernel_size=3,
                stride=stride,
                padding=1,
                groups=hidden_dim,
                bias=False
            ),
            nn.BatchNorm2d(hidden_dim),
            nn.SiLU()
        ])

        layers.append(SEBlock(hidden_dim))

        # Projection
        layers.extend([
            nn.Conv2d(hidden_dim, out_channels, 1, bias=False),
            nn.BatchNorm2d(out_channels)
        ])

        self.block = nn.Sequential(*layers)

    def forward(self, x):
        if self.use_residual:
            return x + self.block(x)
        return self.block(x)


class EfficientNet(nn.Module):

    def __init__(self,
                 in_channels=3,
                 num_classes=10):

        super().__init__()

        self.stem = nn.Sequential(
            nn.Conv2d(
                in_channels,
                32,
                kernel_size=3,
                stride=1,
                padding=1,
                bias=False
            ),
            nn.BatchNorm2d(32),
            nn.SiLU()
        )

        self.features = nn.Sequential(

            MBConv(32, 16, expand_ratio=1, stride=1),

            MBConv(16, 24, expand_ratio=6, stride=1),
            MBConv(24, 24, expand_ratio=6, stride=1),

            MBConv(24, 40, expand_ratio=6, stride=2),
            MBConv(40, 40, expand_ratio=6, stride=1),

            MBConv(40, 80, expand_ratio=6, stride=2),
            MBConv(80, 80, expand_ratio=6, stride=1),

            MBConv(80, 112, expand_ratio=6, stride=1),
            MBConv(112, 112, expand_ratio=6, stride=1),

            MBConv(112, 192, expand_ratio=6, stride=2),
            MBConv(192, 192, expand_ratio=6, stride=1),

            MBConv(192, 320, expand_ratio=6, stride=1),
        )

        self.head = nn.Sequential(
            nn.Conv2d(320, 1280, kernel_size=1, bias=False),
            nn.BatchNorm2d(1280),
            nn.SiLU(),

            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),

            nn.Dropout(0.2),
            nn.Linear(1280, num_classes)
        )

    def forward(self, x):
        x = self.stem(x)
        x = self.features(x)
        x = self.head(x)
        return x
```
